## 5. Listas y cálculo básico

Las listas permiten almacenar varios valores. Por ejemplo, se pueden guardar varias mediciones de temperatura tomadas durante una prueba.

```python
temperaturas_C = [35.2, 36.8, 38.1, 40.5, 43.0, 45.6]

print("Temperaturas:",temperaturas_C )
print("Número de mediciones:", len(temperaturas_C) )
print("Temperatura mínima:",min(temperaturas_C) )
print("Temperatura máxima:",max(temperaturas_C) )
print("Temperatura promedio:",sum (temperaturas_C)/len(temperaturas_C) )
```

## 6. Funciones simples

Una función permite reutilizar una operación. En análisis de datos, las funciones ayudan a ordenar el código y evitar repeticiones.

```python
def calcular_potencia(voltaje, corriente):
    potencia = voltaje * corriente
    return potencia

calcular_potencia(12, 1.8)
```

```python
def clasificar_tempertura(temp):
    if temp < 40:
      return "Normal"
    elif temp<45:
      return "Precaucion"
    else:
      return "Alerta"

for temp in temperaturas_C:
    print(clasificar_tempertura(temp))
```

## 7. Primer acercamiento a una tabla de datos con Pandas

En análisis de datos, normalmente trabajamos con tablas. En Python, una de las estructuras más utilizadas para esto es el `DataFrame` de la librería `pandas`.

```python
datos = {
    "muestra": [1, 2, 3, 4, 5, 6],
    "voltaje_V": [12.0, 12.1, 11.9, 12.2, 12.0, 12.3],
    "corriente_A": [1.6, 1.7, 1.8, 2.0, 2.1, 2.3],
    "temperatura_C": [35.2, 36.8, 38.1, 40.5, 43.0, 45.6]
}

df= pd.DataFrame(datos)
df
```

Agreguemos una nueva columna calculada: potencia eléctrica.

```python
plt.figure(figsize=(8, 4))
plt.plot(df["muestra"], df["potencia"], marker="o")
plt.title("potencia durante la prueba")
plt.xlabel("Muestra")
plt.ylabel("potencia [W]")
plt.grid(True)
plt.show()
```

## Diagnóstico — Parte A: Variables y operaciones

Cree las siguientes variables:

- `nombre_equipo`: nombre de un equipo o sistema mecatrónico.
- `voltaje`: valor numérico en voltios.
- `corriente`: valor numérico en amperios.
- `tiempo_operacion`: tiempo de operación en horas.

Luego calcule:

- `potencia` y `energia`. Sabiendo que la energía es el producto entre potencia y tiempo de operación.

Finalmente, imprima una frase que resuma los resultados.

```python
# TODO: complete el ejercicio de variables y operaciones
nombre_equipo = "Motor DC "
voltaje = 12.0          # voltios
corriente = 1.8         # amperios
tiempo_operacion = 3.5  # horas

# Calcule potencia y energía
potencia = voltaje * corriente          #  (W)
energia = potencia * tiempo_operacion   #  (Wh)

# Imprima los resultados
print(f"El equipo '{nombre_equipo}' operó con un voltaje de {voltaje} V y una corriente de {corriente} A, "
      f"generando una potencia de {potencia:.2f} W. Durante {tiempo_operacion} horas de funcionamiento, "
      f"consumió una energía total de {energia:.2f} Wh.")
```

## Diagnóstico — Parte B: Listas

```python
vibracion = [0.12, 0.15, 0.18, 0.30, 0.28, 0.20, 0.16, 0.14]

# TODO: calcule los valores solicitados
print("Vibracion:",vibracion )
print("Número de mediciones:", len(vibracion ) )
print("Temperatura mínima:",min(vibracion ) )
print("Temperatura máxima:",max(vibracion ) )
print("Temperatura promedio:",sum (vibracion )/len(vibracion ) )
```

## Diagnóstico — Parte C: Condicionales

Cree una función llamada `clasificar_vibracion(valor)` que retorne:

- `"Normal"` si el valor es menor que 0.20.
- `"Revisar"` si el valor está entre 0.20 y 0.29.
- `"Alerta"` si el valor es mayor o igual que 0.30.

Luego aplique la función a cada valor de la lista `vibracion`.

```python
# TODO: cree la función y aplíquela a la lista de vibración
def clasificar_vibracion(valor):
    if valor < 0.20:
        return "Normal"
    elif valor < 0.30:
        return "Revisar"
    else:
        return "Alerta"

for valor in vibracion:
    print(f"Vibración: {valor} -> {clasificar_vibracion(valor)}")
```
