## Parte 1 — Función para convertir valores

Cree una función llamada `convertir_a_float(valor)` que intente convertir un valor a número decimal.

La función debe retornar:

- El número convertido, si la conversión es posible.
- `None`, si el valor no se puede convertir.

````python
# TODO: cree la función convertir_a_float

def convertir_a_float(valor):
    try:
        return float(valor)
    except (ValueError, TypeError):
        return None

# Pruebas sugeridas
print(convertir_a_float("12.5"))   # 12.5
print(convertir_a_float("error"))  # None
print(convertir_a_float(""))       # None
print(convertir_a_float(None))     # None
````

Esta función centraliza la conversión de texto a número, manejando los casos Se creó la función convertir_a_float, que intenta convertir cualquier valor a número decimal usando float(). Si el valor no se puede convertir (por ejemplo, un texto inválido, un string vacío o None), la función captura el error con try/except y retorna None en lugar de detener la ejecución del programa

## Parte 2 — Función para calcular potencia

Cree una función llamada `calcular_potencia(voltaje, corriente)`.

La función debe:

- Retornar `None` si voltaje o corriente son `None`.
- Retornar `None` si voltaje o corriente son negativos.
- Retornar `voltaje * corriente` en los demás casos.

````python
# TODO: cree la función calcular_potencia

def calcular_potencia(voltaje, corriente):
    if voltaje is None or corriente is None:
        return None
    if voltaje < 0 or corriente < 0:
        return None
    return voltaje * corriente

# Pruebas sugeridas
print(calcular_potencia(12, 2))    # 24
print(calcular_potencia(None, 2))  # None
print(calcular_potencia(12, -1))   # None
````

La función calcular_potencia recibe el voltaje y la corriente y calcula la potencia multiplicándolos. Antes de calcular primero se revisa si alguno de los dos valores es None porque si se intenta comparar None con un número Python da error. Después se valida que no sean negativos ya que en este contexto físico no tiene sentido tener voltaje o corriente negativos. Si pasa ambas validaciones se retorna el producto voltaje * corriente

## Parte 3 — Función para clasificar temperatura

Cree una función llamada `clasificar_temperatura(temp)` con los siguientes criterios:

- Si `temp` es `None`: `"Dato inválido"`
- Si `temp < 40`: `"Normal"`
- Si `40 <= temp < 50`: `"Precaución"`
- Si `temp >= 50`: `"Alerta"`

````python
# TODO: cree la función clasificar_temperatura

def clasificar_temperatura(temp):
    if temp is None:
        return "Dato inválido"
    elif temp < 40:
        return "Normal"
    elif temp < 50:
        return "Precaución"
    else:
        return "Alerta"

# Pruebas sugeridas
print(clasificar_temperatura(35))   # Normal
print(clasificar_temperatura(45))   # Precaución
print(clasificar_temperatura(52))   # Alerta
print(clasificar_temperatura(None)) # Dato inválido
````

La función clasificar_temperatura clasifica la temperatura de un motor en tres categorías según su valor. Primero se valida si el dato es None en cuyo caso se retorna "Dato inválido" para evitar comparar None con números. Luego si la temperatura es menor a 40 se considera "Normal", si está entre 40 y 50 se considera "Precaución", y si es mayor o igual a 50 se considera "Alerta". No fue necesario escribir el rango completo en el segundo elif (por ejemplo 40 <= temp < 50), porque al llegar ahí ya se sabe que la temperatura no es menor a 40.

## Parte 4 — Limpieza y procesamiento de registros

Procese la lista `registros_crudos` y cree una nueva lista llamada `registros_limpios`.

Cada registro limpio debe contener:

- `id`
- `voltaje`
- `corriente`
- `temperatura`
- `potencia`
- `estado_temperatura`

Use las funciones creadas.

````python
# Datos de prueba (simulan lecturas crudas de sensores de motores)
registros_crudos = [
    {"id": "M01", "voltaje": "12.5", "corriente": "2.0", "temperatura": "35"},
    {"id": "M02", "voltaje": "24",   "corriente": "1.8", "temperatura": "45"},
    {"id": "M03", "voltaje": "error", "corriente": "1.2", "temperatura": "52"},
    {"id": "M04", "voltaje": "18",   "corriente": "",     "temperatura": "38"},
    {"id": "M05", "voltaje": "-5",   "corriente": "2.5",  "temperatura": "41"},
    {"id": "M06", "voltaje": "9.6",  "corriente": "0.9",  "temperatura": None},
    {"id": "M07", "voltaje": "15",   "corriente": "3.0",  "temperatura": "55"},
    {"id": "M08", "voltaje": None,   "corriente": "1.5",  "temperatura": "30"},
]
````

````python
# TODO: procese los registros

registros_limpios = []

for r in registros_crudos:
    voltaje = convertir_a_float(r["voltaje"])
    corriente = convertir_a_float(r["corriente"])
    temperatura = convertir_a_float(r["temperatura"])

    potencia = calcular_potencia(voltaje, corriente)
    estado_temperatura = clasificar_temperatura(temperatura)

    registro_limpio = {
        "id": r["id"],
        "voltaje": voltaje,
        "corriente": corriente,
        "temperatura": temperatura,
        "potencia": potencia,
        "estado_temperatura": estado_temperatura,
    }

    registros_limpios.append(registro_limpio)

registros_limpios
````

Se recorrió registros_crudos con un ciclo for, y para cada registro se usaron las funciones creadas anteriormente: primero convertir_a_float para pasar voltaje, corriente y temperatura de texto a número (o None si el dato era inválido), luego calcular_potencia con esos valores ya convertidos y clasificar_temperatura con la temperatura. Con esa información se armó un nuevo diccionario registro_limpio por cada motor y se fue agregando a la lista registros_limpios. Esta parte no tuvo que repetir ninguna validación porque toda esa lógica ya estaba resuelta en las funciones de las Partes 1 a 3

## Parte 5 — Comprensión de listas

Use comprensión de listas para obtener:

1. Lista de potencias válidas.
2. Lista de temperaturas válidas.
3. Lista de motores en alerta.

````python
# TODO: use comprensión de listas

# 1. Lista de potencias válidas (descarta los None)
potencias_validas = [r["potencia"] for r in registros_limpios if r["potencia"] is not None]

# 2. Lista de temperaturas válidas (descarta los None)
temperaturas_validas = [r["temperatura"] for r in registros_limpios if r["temperatura"] is not None]

# 3. Lista de motores en alerta (ids donde estado_temperatura == "Alerta")
motores_alerta = [r["id"] for r in registros_limpios if r["estado_temperatura"] == "Alerta"]

print(potencias_validas)
print(temperaturas_validas)
print(motores_alerta)
````

Se usaron comprensiones de listas para extraer información específica de registros_limpios de forma más compacta que un ciclo for tradicional. potencias_validas y temperaturas_validas filtran los registros descartando los valores None, ya que esos datos no se pueden usar en cálculos posteriores, motores_alerta en cambio filtra por el id de los motores cuyo estado_temperatura sea "Alerta", para identificar rápidamente cuáles requieren atención

## Parte 6 — Comprensión de diccionarios

Cree un diccionario llamado `estado_por_motor` donde:

- La clave sea el `id` del motor.
- El valor sea el `estado_temperatura`.

````python
# TODO: cree el diccionario estado_por_motor

estado_por_motor = {r["id"]: r["estado_temperatura"] for r in registros_limpios}

estado_por_motor
````

Se usó una comprensión de diccionarios para construir estado_por_motor, donde cada clave es el id del motor y cada valor es su estado_temperatura. A diferencia de las listas de la Parte 5, aquí no se filtró ningún registro, porque interesa tener el estado de todos los motores, incluidos los que tienen "Dato inválido" eso permite identificar de un vistazo qué motores tienen problemas de sensor, no solo cuáles están en alerta

## Parte 7 — Resumen automático

Cree una función llamada `generar_resumen(registros)` que retorne un diccionario con:

- `total_registros`
- `registros_validos_potencia`
- `potencia_promedio`
- `temperatura_promedio`
- `cantidad_alertas`
- `cantidad_precauciones`
- `cantidad_normales`

````python
# TODO: cree la función generar_resumen

def generar_resumen(registros):
    total_registros = len(registros)

    potencias_validas = [r["potencia"] for r in registros if r["potencia"] is not None]
    temperaturas_validas = [r["temperatura"] for r in registros if r["temperatura"] is not None]

    registros_validos_potencia = len(potencias_validas)

    potencia_promedio = sum(potencias_validas) / len(potencias_validas) if potencias_validas else None
    temperatura_promedio = sum(temperaturas_validas) / len(temperaturas_validas) if temperaturas_validas else None

    cantidad_alertas = sum(1 for r in registros if r["estado_temperatura"] == "Alerta")
    cantidad_precauciones = sum(1 for r in registros if r["estado_temperatura"] == "Precaución")
    cantidad_normales = sum(1 for r in registros if r["estado_temperatura"] == "Normal")

    return {
        "total_registros": total_registros,
        "registros_validos_potencia": registros_validos_potencia,
        "potencia_promedio": potencia_promedio,
        "temperatura_promedio": temperatura_promedio,
        "cantidad_alertas": cantidad_alertas,
        "cantidad_precauciones": cantidad_precauciones,
        "cantidad_normales": cantidad_normales,
    }

resumen = generar_resumen(registros_limpios)
resumen
````

La función generar_resumen recibe la lista de registros y calcula varias métricas: el total de registros, cuántos tienen potencia válida, el promedio de potencia y de temperatura (usando solo los valores no nulos, para no dividir por cero si todos fueran inválidos), y la cantidad de motores en cada estado (Alerta, Precaución, Normal). Todo se retorna en un solo diccionario con los resultados.
